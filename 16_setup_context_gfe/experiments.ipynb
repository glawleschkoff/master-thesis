```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import xarray as xr
from simulations import run_simulation_1, run_simulation_2, run_simulation_3
```

Simulation 1

```python
#run_simulation_1()
```

```python
xdata = xr.open_dataarray('artifacts/simulation_1.nc').load()
xdata = xdata.loc[:, :, 1]
df_data = xdata.T.to_pandas()
df_data.index = df_data.index.round(1)
df_data.columns = df_data.columns.round(1)

plt.figure(figsize=(12, 8))

ax = sns.heatmap(df_data,
                annot=True,
                cmap="viridis",
                fmt=".3f",
                annot_kws={'size': 15},
                vmin=0.0,
                vmax=0.4)

ax.xaxis.tick_top()
ax.tick_params(labelsize=12)

ax.xaxis.set_label_position('top')

plt.title("Context GFE Probability to Cue at t=1", pad=10, fontsize=18) 
plt.ylabel(rf"$\alpha$", fontsize=22)
plt.xlabel("c", fontsize=22)

plt.tight_layout()
plt.savefig('artifacts/simulation_1_cue.png', dpi=300, bbox_inches='tight')
plt.show()
```
