```python
# 1. Daten laden
xdata = xr.open_dataarray('artifacts/simulation_4.nc').load()
iterations = xdata.coords['iteration']

# 2. Parameter-Listen definieren
c_values = np.arange(0, 11)
alpha_values = np.arange(0, 1.1, 0.1)

# 3. Figure mit 3x3 Subplots erstellen
fig, axes = plt.subplots(nrows=11, ncols=11, figsize=(40, 30))

# Farben definieren
color_fe = 'tab:orange'
color_q = 'tab:blue'

# 4. Über die Zeilen (c) und Spalten (alpha) iterieren
for i, c in enumerate(c_values):
    for j, alpha in enumerate(alpha_values):
        
        # Den aktuellen Subplot auswählen
        ax1 = axes[i, j]
        
        # Daten für diese spezifische Kombination extrahieren
        q_s0_trace = xdata.sel(c=c, alpha=alpha, value=0)
        free_energy_trace = xdata.sel(c=c, alpha=alpha, value=1)
        
        # --- Linke y-Achse: Free Energy ---
        ax1.plot(iterations, free_energy_trace, label="Free Energy", color=color_fe)
        ax1.tick_params(axis='y', labelcolor=color_fe)
        
        # Dynamische Limits berechnen
        fe_max = free_energy_trace.max().item()
        fe_min = free_energy_trace.min().item()
        
        # NEUE LOGIK: Toleranz für Gleitkomma-Ungenauigkeiten nutzen!
        # Wenn die Schwankung fast null ist, behandeln wir es als flache Linie.
        if abs(fe_max - fe_min) < 1e-6:
            fe_min = fe_max - 2
            
        # Puffer berechnen (5 % der Gesamtschwankung)
        span = fe_max - fe_min
        puffer = 0.05 * span
        
        # Limit symmetrisch setzen: Unten abziehen, oben addieren
        ax1.set_ylim(bottom=fe_min - puffer, top=fe_max + puffer)
        
        # x-Label nur für die unterste Reihe zeichnen
        if i == len(c_values) - 1:  # <-- Korrigiert: Jetzt immer ganz unten!
            ax1.set_xlabel("Iteration", fontsize=24)
            
        # y-Label nur für die linke Spalte zeichnen
        if j == 0:
            ax1.set_ylabel("Free Energy", color=color_fe, fontsize=24)
            
        ax1.set_title(rf"$c={c}$, $\alpha={alpha:.1f}$", fontsize=20)

        # --- Rechte y-Achse: q(c=0) ---
        ax2 = ax1.twinx()
        ax2.plot(iterations, q_s0_trace, label="q(c=0)", color=color_q)
        ax2.tick_params(axis='y', labelcolor=color_q)
        ax2.set_ylim(0, 1)
        
        # y-Label nur für die ganz rechte Spalte zeichnen
        if j == len(alpha_values) - 1:  # <-- Korrigiert: Jetzt immer ganz rechts!
            ax2.set_ylabel("q(c=0)", color=color_q, fontsize=24)

# 5. EINE gemeinsame Legende für alle 9 Plots erstellen
lines_1, labels_1 = ax1.get_legend_handles_labels()
lines_2, labels_2 = ax2.get_legend_handles_labels()
fig.legend(
    lines_1 + lines_2, 
    labels_1 + labels_2, 
    loc='upper center', 
    ncol=2,
    fontsize=24, 
    bbox_to_anchor=(0.5, 1.02)
)
fig.suptitle(
    "Context BFE Agent (1000 Iterations)", 
    fontsize=36, 
    fontweight='bold', 
    y=1.05  # Schiebt den Titel über die Legende (die bei 1.05 liegt)
)

# Layout optimieren
plt.tight_layout()

# Plot speichern und anzeigen
plt.savefig('artifacts/simulation_4.png', dpi=400, bbox_inches='tight')
plt.show()
```
