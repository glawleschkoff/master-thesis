```python
from agent_gfe import GFEAgent
from agent_gfe_param import generate_gfe_params
from environment import Environment
from environment_param import a, b, d

time_horizon = 3

(A, B, C, D, U) = generate_gfe_params(alpha=0.9, c=2, beta=1)
agent = GFEAgent(A, B, C, D, U, time_horizon)
environment = Environment(a, b, d)

observation = environment.generate_observation()
print('Observation: ' + str(observation))
agent.observe(observation)
agent.infer()
action = agent.act()
print('Action: ' + str(action))
environment.act_upon(action)

observation = environment.generate_observation()
print('Observation: ' + str(observation))
agent.observe(observation)
agent.infer()
action = agent.act()
print('Action: ' + str(action))
environment.act_upon(action)

observation = environment.generate_observation()
print('Observation: ' + str(observation))

agent.print_beliefs()
```
