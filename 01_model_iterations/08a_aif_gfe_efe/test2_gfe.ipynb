```python
from agent_gfe import GFEAgent
from agent_gfe_param import generate_gfe_params
from environment import Environment
from environment_param import a, b, d

time_horizon = 3

(A, B, C, D, U) = generate_gfe_params(c=2, alpha=0.9)
agent = GFEAgent(A, B, C, D, U, time_horizon)
environment = Environment(a, b, d)
```

##### (1) Agent's beliefs after thinking

```python
agent.infer()
agent.print_beliefs()
```

##### (2) Environment generates observation ...

```python
observation = environment.generate_observation()
print('Observation: ' + str(observation))
```

##### (3) Agent observes and infers (observation is not informative)...

```python
agent.observe(observation)
agent.infer()
agent.print_beliefs()
```

##### (4) Agent acts ...

```python
action = agent.act(1)
environment.act_upon(action)
print('Action: ' + str(action))
```

##### (5) Environment generates new observation ... (4 is cue left, 5 is cue right, only 10 and 14 is reward)

```python
observation = environment.generate_observation()
print('Observation: ' + str(observation))
```

##### (6) Agent observes and infers ...

```python
agent.observe(observation)
agent.infer()
agent.print_beliefs()
```

##### (7) Agent acts ...

```python
action = agent.act()
environment.act_upon(action)
print('Action: ' + str(action))
```

##### (8) Environment generates final observation ... (only 10 and 14 is reward)

```python
observation = environment.generate_observation()
print('Observation: ' + str(observation))
```

##### (9) Agent observes and infers ...

```python
agent.observe(observation)
agent.infer()
agent.print_beliefs()
```
